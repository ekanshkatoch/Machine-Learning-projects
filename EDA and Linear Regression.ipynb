```python
print("Numerical variables:")
display(df.describe())

print("\nCategorical variables:")
display(df.describe(include='object'))
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

print("Mean:")
print(df[numeric_cols].mean())

print("\nMedian:")
print(df[numeric_cols].median())

print("\nStandard Deviation:")
print(df[numeric_cols].std())

print("\nMinimum:")
print(df[numeric_cols].min())

print("\nMaximum:")
print(df[numeric_cols].max())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols].hist(
    figsize=(15, 12),
    bins=20,
    edgecolor='black'
)

plt.suptitle("Distribution of Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="annual_income",
    y="loan_amount",
    hue="loan_paid_back"
)

plt.title("Annual Income vs Loan Amount by Loan Paid Back")
plt.show()
```

```python
plt.figure(figsize=(15, 8))

sns.boxplot(data=df[numeric_cols])

plt.title("Box Plot of Numerical Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
correlation = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

sns.pairplot(
    df[numeric_cols],
    diag_kind="hist"
)

plt.suptitle("Pair Plot of Numerical Variables", y=1.02)
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("/kaggle/input/datasets/deeplumiere/load-pred-dataset/original_dataset.csv")

df = df[["annual_income", "loan_amount"]].dropna()

X = df[["annual_income"]]
y = df["loan_amount"]

model = LinearRegression()

model.fit(X, y)

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Regression Equation:")
print("loan_amount =", model.intercept_, "+", model.coef_[0], "* annual_income")
```
